Now let's learn apply(), another important Pandas concept.

Your task:

Create a new column called Profit_Status where:

Profit > 0 → "Profit"
Profit < 0 → "Loss"
Profit = 0 → "No Profit No Loss"

```python
df1['Profit_Status']=df1.apply(lambda x: 'Profit' if x['Profit']>0 else ('Loss' if x['Profit']<0 else('No loss/No profit')),axis=1)
df1['Profit_Status'].value_counts()
```

Next challenge: apply() with a custom function

Create a new column called Sales_Category:

Sales < 100 → "Low"
Sales 100–500 → "Medium"
Sales > 500 → "High"

```python
df1['Sales_Category']=df1.apply(lambda x: 'Low' if x['Sales']<100 else ('Medium' if x['Sales']<500 else('High')),axis=1)
df1['Sales_Category'].value_counts()
```

Now use your existing Customer_Total_Sales column and answer:

How many unique customers fall into Low, Medium, and High total-sales categories?

```python
df1.apply(lambda x: 'Low' if x['Customer_Total_Sales']<100 else ('Medium' if x['Customer_Total_Sales']<500 else('High')),axis=1).value_counts()
```

We'll learn it in a practical order:

Bar chart → Category Sales

Bar chart → Category Profit

Line chart → Year/Month Sales trend

Line chart → Year/Month Profit trend

Bar chart → Top 10 customers/products

Scatter plot → Discount vs Profit

Heatmap → Correlation

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df1,x='Category',y='Sales',estimator='sum')
plt.title('Category Sales')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(data=df1,x='Category',y='Profit',estimator='sum')
plt.title('Category Profit')
plt.show()
```

```python
monthly_sales_trend = df1.groupby(['Year', 'Month'])['Sales'].sum().reset_index()

plt.figure(figsize=(12, 6))

for year in monthly_sales_trend['Year'].unique():
    data = monthly_sales_trend[monthly_sales_trend['Year'] == year]
    plt.plot(data['Month'], data['Sales'], marker='o', label=str(year))

plt.title('Monthly Sales Trend by Year')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.legend(title='Year')
plt.xticks(rotation=45)
plt.show()
```

```python
monthly_profit_trend = df1.groupby(['Year', 'Month'])['Profit'].sum().reset_index()

plt.figure(figsize=(12, 6))

for year in monthly_profit_trend['Year'].unique():
    data = monthly_profit_trend[monthly_profit_trend['Year'] == year]
    plt.plot(data['Month'], data['Profit'], marker='o', label=str(year))

plt.title('Monthly Sales Trend by Year')
plt.xlabel('Month')
plt.ylabel('Total Profit')
plt.legend(title='Year')
plt.xticks(rotation=45)
plt.show()
```

```python
top_customers=df1.groupby('Customer Name')['Customer_Total_Sales'].max().nlargest(10)
sns.barplot(x=top_customers,y=top_customers.index)
plt.title('Top 10 Customers')
plt.xlabel('Total Sales')
plt.ylabel('Customer Name')
plt.show()
```

```python
top_product=df1.groupby('Product Name')['Sales'].sum().nlargest(10)
sns.barplot(x=top_product,y=top_product.index)
plt.title('Top 10 Products')
plt.xlabel('Total Sales')
plt.ylabel('Product Name')
plt.show()
```

```python
sns.scatterplot(data=df1,x='Discount',y='Profit')
plt.title('Discount vs Profit')
plt.show()
```

```python
temp=df1[['Sales','Quantity','Profit','Discount']].corr()
sns.heatmap(temp)
plt.title("Heat map")
plt.show()
```
